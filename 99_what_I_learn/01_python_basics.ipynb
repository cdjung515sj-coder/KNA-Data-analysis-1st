# 01 · Python 기본 문법과 자료형

출력 → 변수 → 자료형 → 문자열 → 컬렉션 순서로 읽습니다. 같은 결과를 만드는 문법을 비교해 이후 라이브러리 코드도 해석할 수 있게 합니다.

> 위에서 아래로 실행하세요. 예제 데이터는 노트북 안에서 만듭니다. 코드 셀 아래의 출력으로 결과를 확인하고, 실제 데이터에서는 열 이름·단위·기간을 먼저 확인하세요.

## 출력과 변수

**코드 → 코드 개념**: `print`는 값을 보여 주고 `=`는 값을 이름에 저장한다. `==`는 같은지 비교한다.

**코드 사용법**: 값을 바꾼 뒤 아래 셀을 다시 실행해 출력과 불리언 결과를 비교한다.

```python
machine = "A-01"
temperature = 78.5
print(machine, temperature)
print(f"{machine}: {temperature:.1f} °C")
print(temperature == 78.5, type(temperature))
```

**같은 결과를 얻는 방법과 선택 이유**

- `print(machine, temperature)`는 빠른 점검에 편하다. `f"..."`는 단위와 소수 자리까지 표시하는 보고 문구에 좋다.
- `type()`은 계산 결과가 문자열인지 숫자인지 헷갈릴 때 확인한다. `78.5`와 `"78.5"`는 다르다.

## 연산자와 형 변환

**코드 → 코드 개념**: `/`는 실수 나눗셈, `//`는 몫, `%`는 나머지다. `int`·`float`는 자료형을 바꾼다.

**코드 사용법**: 문자열로 읽은 센서 값을 숫자로 바꾼 뒤 평균을 낸다.

```python
raw = ["70", "80", "90"]
numbers = [float(x) for x in raw]
print(sum(numbers) / len(numbers))
print(7 / 2, 7 // 2, 7 % 2)
print(2 ** 3, 10 >= 8)
```

**같은 결과를 얻는 방법과 선택 이유**

- `float(x)`는 소수가 있는 측정값에, `int(x)`는 정수 코드에 쓴다. 숫자가 아닌 문자열은 오류가 나므로 실제 파일에는 변환 실패 처리도 필요하다.
- 평균을 직접 계산하는 방식은 원리를 보기에 좋고, Pandas의 `.mean()`은 표의 결측을 처리하며 열별 집계에 편하다.

## 문자열

**코드 → 코드 개념**: 문자열은 순서가 있는 문자들의 모음이다. 인덱스·슬라이스·메서드로 가공한다.

**코드 사용법**: 태그의 양끝 공백을 없애고 구분자로 분리한다.

```python
tag = "  LINE-A-TEMP  "
clean = tag.strip()
print(clean, clean.lower(), clean.split("-"))
print(clean[:4], clean.replace("TEMP", "VIB"))
```

**같은 결과를 얻는 방법과 선택 이유**

- `strip()`은 양끝 공백만 제거한다. 중간 공백까지 없애려면 `replace()`를 쓰되 식별자의 의미를 바꾸지 않는지 본다.
- `split('-')`은 단순 구분자에 적합하다. 형식이 여러 가지면 규칙 검증이나 정규표현식이 필요하다.

## 리스트와 슬라이싱

**코드 → 코드 개념**: 리스트는 순서가 있고 수정 가능하다. `a:b`는 a부터 b 직전까지 선택한다.

**코드 사용법**: 측정값 추가·정렬·부분 선택을 비교한다.

```python
values = [75, 70, 90]
values.append(80)
print(values[0], values[-1], values[1:3])
print(sorted(values), values)  # 원본 보존
values.sort()                  # 원본 변경
print(values)
```

**같은 결과를 얻는 방법과 선택 이유**

- `sorted(values)`는 새 목록이 필요할 때, `values.sort()`는 원본을 바꿔도 될 때 쓴다. `sort()`의 반환값은 `None`이다.
- `values[1:3]`과 `values[:2]`는 같은 결과가 될 수 있지만 전자는 시작·끝 위치를 명시하고 후자는 맨 앞부터라는 의도를 드러낸다.

## 튜플·집합·딕셔너리

**코드 → 코드 개념**: 튜플은 고정된 순서, 집합은 중복 없는 원소, 딕셔너리는 키와 값의 대응이다.

**코드 사용법**: 설비 코드와 측정값을 목적에 맞는 자료형에 담는다.

```python
pair = ("A", 80)
machines = ["A", "B", "A"]
record = {"machine": "A", "temperature": 80}
print(pair[0], set(machines), record["temperature"])
print(list(dict.fromkeys(machines)))  # 처음 등장한 순서대로 중복 제거
```

**같은 결과를 얻는 방법과 선택 이유**

- `set(machines)`는 중복을 없애지만 순서를 보존해야 하는 출력에는 적합하지 않다. `dict.fromkeys`는 처음 등장 순서를 유지한다.
- 이름으로 값을 찾는 데이터는 딕셔너리, 고정된 두 값의 묶음은 튜플이 자연스럽다.

## 입력과 예외

**코드 → 코드 개념**: `input()`은 문자열을 돌려준다. `try/except`는 변환 실패를 처리한다.

**코드 사용법**: 아래 함수에 정상 입력과 잘못된 입력을 각각 넣어 본다.

```python
def parse_temperature(text):
    try:
        return float(text.strip())
    except ValueError:
        return None

print(parse_temperature(" 82.5 "), parse_temperature("error"))
```

**같은 결과를 얻는 방법과 선택 이유**

- 작은 함수로 분리하면 대화형 `input()` 없이도 테스트할 수 있다. 직접 `float(input())`을 쓰면 간단하지만 잘못된 입력에 실행이 멈춘다.
- 실패를 `None`으로 돌려주면 0°C와 결측을 구분할 수 있다.
