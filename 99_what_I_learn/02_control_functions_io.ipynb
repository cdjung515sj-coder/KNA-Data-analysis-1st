## 조건문

**코드 → 코드 개념**: `if/elif/else`는 위에서부터 검사해 처음 맞는 분기를 실행한다.

**코드 사용법**: 진동값을 세 수준으로 분류한다.

```python
vibration = 3.4
if vibration >= 4.0:
    state = "danger"
elif vibration >= 3.0:
    state = "watch"
else:
    state = "normal"
print(state)
print("high" if vibration >= 3.0 else "low")
```

**같은 결과를 얻는 방법과 선택 이유**

- 여러 단계면 `if/elif/else`가 읽기 쉽다. 두 결과 중 하나를 한 줄에서 선택하면 조건부 표현식이 간결하다.
- `>= 4.0`을 먼저 검사해야 4 이상이 `watch`로 분류되지 않는다. 기준값은 설비 규격에 맞춰 정한다.

## 반복과 누적

**코드 → 코드 개념**: `for`는 각 값을 순회하고 `sum`은 합계를 계산한다.

**코드 사용법**: 같은 초과 건수를 두 방식으로 얻는다.

```python
values = [2.1, 3.5, 4.2, 2.8]
count = 0
for value in values:
    if value >= 3.0:
        count += 1
count2 = sum(value >= 3.0 for value in values)
print(count, count2)
assert count == count2
```

**같은 결과를 얻는 방법과 선택 이유**

- `for`와 누적 변수는 계산 과정을 추적하거나 복잡한 조건을 넣을 때 좋다. `sum(조건 for ...)`는 단순 건수 계산에 좋다.
- 표 데이터에서는 `(df['vibration'] >= 3).sum()`이 행별 연산을 한 번에 처리한다.

## while, break, continue

**코드 → 코드 개념**: `while`은 조건이 참인 동안 반복한다. `break`는 종료, `continue`는 다음 회차로 이동한다.

**코드 사용법**: 처음으로 위험 기준을 넘는 위치를 찾는다.

```python
values = [None, 2.1, 3.5, 4.2]
position = 0
while position < len(values):
    value = values[position]
    position += 1
    if value is None:
        continue
    if value >= 4.0:
        break
print(position - 1, value)
```

**같은 결과를 얻는 방법과 선택 이유**

- 인덱스를 직접 조절해야 할 때 `while`을 쓴다. 단순 순회는 `for index, value in enumerate(values)`가 종료 조건을 잊을 위험이 적다.
- 검색 결과가 없을 수 있으면 반복 종료 후 별도 상태를 확인해야 한다.

## 함수, 매개변수, 반환

**코드 → 코드 개념**: 함수는 반복되는 규칙을 이름으로 묶고 `return`으로 계산 결과를 돌려준다.

**코드 사용법**: 비어 있는 입력을 처리하는 비율 함수를 만들고 재사용한다.

```python
def exceedance_rate(values, threshold=3.0):
    valid = [x for x in values if x is not None]
    if not valid:
        return None
    return sum(x >= threshold for x in valid) / len(valid)

print(exceedance_rate([2.1, 3.5, None, 4.2]))
print(exceedance_rate([2.1, 3.5], threshold=4.0))
```

**같은 결과를 얻는 방법과 선택 이유**

- 기본값은 자주 쓰는 기준을 제공하고 호출 시 바꿀 수 있다. 고정 기준을 함수 내부에 박아 두면 설비마다 적용하기 어렵다.
- `print`는 보여 주기만 하고, `return`은 이후 계산에 재사용할 값을 만든다.

## 모듈과 파일 입출력

**코드 → 코드 개념**: `import`는 표준 라이브러리 기능을 가져온다. `with open`은 파일을 자동으로 닫는다.

**코드 사용법**: 먼저 메모리 안의 파일로 쓰기·읽기를 연습한다. 실제 파일에는 `Path`를 쓴다.

```python
from io import StringIO
from pathlib import Path

buffer = StringIO()
buffer.write("A,2.1\nB,3.4\n")
buffer.seek(0)
print(buffer.read().splitlines())
print(Path("data") / "notes.txt")
```

**같은 결과를 얻는 방법과 선택 이유**

- `StringIO`는 파일을 만들지 않고 읽기·쓰기를 연습하거나 테스트할 때 좋다. 실제 작은 텍스트 파일은 `Path('notes.txt').write_text(text, encoding='utf-8')`와 `read_text(...)`로 처리한다. 큰 파일은 `with open(...)`으로 줄마다 읽는다.
- 표 형태 CSV는 문자열 분리보다 `pd.read_csv`가 따옴표·결측·열 이름 처리를 잘한다.
