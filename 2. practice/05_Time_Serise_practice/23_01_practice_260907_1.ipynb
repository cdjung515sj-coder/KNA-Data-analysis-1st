## 실습 2. 단일 센서 이동평균 추적

## 목표
창 크기 10 이동평균을 구해 원본 위에 겹쳐 추세를 읽기
### 단계
- 센서에 창 크기 10 이동평균을 새 열로 담기
- 원본을 옅게, 이동평균을 진하게 겹쳐 그리기

### 예상 결과
들쭉날쭉한 원본 위로 매끈한 이동평균 추세선

```python
# 실습 2 코드
# unit2 = df[df["설비번호"] == 2].copy().sort_values("시퀀스").reset_index(drop=True)

# unit2["ma10"] = unit2["컨베이어전류"]
# unit2["ma10"] = unit2["컨베이어전류"].rolling(window=10)  # 오류 발생
unit2["ma10"] = unit2["회전A전류"].rolling(window=10).mean()

unit2.head()
unit2.tail()

plt.plot(unit2["시퀀스"], unit2["회전A전류"], color = "lightgray", label = "raw")
plt.plot(unit2["시퀀스"], unit2["ma10"], color = "red", label = "MA(10)")
plt.title("설비 2 - 회전A전류")
plt.legend()
plt.show()
```

## 실습 4. 이동평균·이동표준편차 동시 추적
### 목표
이동평균과 이동표준편차를 함께 계산해 추세와 변동성을 동시에 추적
### 단계
- 창 크기 10으로 이동평균과 이동표준편차를 구하기
- 두 통계를 위아래 두 칸으로 나눠 그리기
- 초반과 후반의 변동성 평균을 비교
### 예상 결과
후반 변동성이 초반의 약 2배(5.5 → 9.5)로 커짐

```python
# 실습 4 코드

# 이동평균
# 이동표준편차
unit2["std10"] = unit2["회전A전류"].rolling(window=10).std()
# unit2.tail()


# 초반 변동성
early = unit2["std10"].iloc[10:50].mean()
late = unit2["std10"].iloc[150:].mean()
print("초반 변동성: ", round(early,4))
print("후반 변동성: ", round(late,4))

# 두 그래프를 subplot으로 위 아래 배치
# plt.plot(unit2["시퀀스"], unit2["ma10"], color = "blue", label = "MA(10)")
plt.plot(unit2["시퀀스"], unit2["std10"], color = "red", label = "STD(10)")
plt.show()

plt.subplot(2, 1, 1)
plt.plot(unit2["시퀀스"], unit2["ma10"], color="blue", label="MA(10)")
plt.title("이동평균 (MA10)")
plt.ylabel("회전A전류")
plt.legend()

plt.subplot(2, 1, 2)
plt.plot(unit2["시퀀스"], unit2["std10"], color="red", label="STD(10)")
plt.title("이동표준편차 (STD10)")
plt.xlabel("시퀀스")
plt.ylabel("변동성(표준편차)")
plt.legend()

plt.tight_layout()
plt.show()
```

## 실습 5. 변화량·변화율 계산과 시각화
### 목표
변화량과 변화율로 급변을 포착하고 이동평균 후 변화량으로 잡음을 줄이기
### 단계
- 직전 대비 변화량과 변화율을 각각 구하기
- 앞부분 값을 확인해 첫 값이 비는 이유 이해
- 이동평균을 먼저 낸 뒤 변화량을 구해 잡음이 줄어듦을 확인
### 예상 결과
변화량 앞부분 -0.43·3.07…, 첫 값은 비교 대상이 없어 빈 값

```python
# 실습 5 코드

# 이동평균

# 변화율
unit2["diff"] = unit2['회전A전류'].diff()
# 0	2	1	0.0	0.0496	0.7037	0.0	0.0	NaN	NaN	NaN	NaN	NaN
# 자기 앞선 대상이 없으니까 계산 못하니까 NaN으로 나오는거임
# 빈 부분처럼 나옴 

# unit2["diff"].head() 

#변화량
unit2["pct"] = unit2['회전A전류'].pct_change()

# 이동평균 변화량
unit2["ma10_diff"] = unit2["ma10"].diff()

unit2.head()

# 위 항목들을 그래프로
plt.figure(figsize=(10, 10))
# 1번 칸: 이동평균
plt.subplot(4, 1, 1)
plt.plot(unit2["시퀀스"], unit2["ma10"], color="blue", label="MA(10)")
plt.title("이동평균 (MA10)")
plt.legend()


# 2번 칸: 이동표준편차
plt.subplot(4, 1, 2)
plt.plot(unit2["시퀀스"], unit2["std10"], color="orange", label="STD(10)")
plt.title("이동표준편차 (STD10)")
plt.legend()


# 3번 칸: 변화량
plt.subplot(4, 1, 3)
plt.plot(unit2["시퀀스"], unit2["ma10_diff"], color="red", label="MA(10) Diff")
plt.title("이동평균 변화량 (MA10 Diff)")
plt.xlabel("시퀀스")
plt.legend()

# 4번 칸: 변화율
plt.subplot(4, 1, 4)
plt.plot(unit2["시퀀스"], unit2["pct"], color="purple", label="PCT")
plt.title("전류 변화율 (PCT)")
plt.xlabel("시퀀스")
plt.legend()

plt.tight_layout()
plt.show()
```

## 실습 6. 단순 변화점 탐지와 구간 표시
### 목표
잔차와 임계값으로 변화점을 찾고 그래프에 세로선으로 표시
### 단계
- 원본에서 이동평균을 뺀 잔차를 구하기
- 잔차 표준편차의 두 배를 임계값으로 정해 변화점 표시
- 변화점 위치를 세로선으로 그래프에 표시
### 예상 결과
임계값 약 16.3, 변화점 13개; 대부분 후반부에 몰림

```python
# 실습 6 코드

roll = unit2["회전A전류"].rolling(10).mean() # 10개씩 묶어 평균값 도출
resid = unit2["회전A전류"] - roll # 잔차 계산


# 잔차 표준편차 2배를 임계값 지정
thr = 3 * resid.std()

# 변화점 찾기
unit2["is_change"] = resid.abs() > thr  # 절대값이 지정된 잔차 보다 크면 True

plt.plot(unit2["시퀀스"], unit2["회전A전류"], color = "lightgray")

for c in unit2.loc[unit2["is_change"], "시퀀스"]:
    plt.axvline(c, color = "lightblue", linestyle = "--", alpha = 0.7)  # 변화점들만 따로 채움
    

plt.show()

```

구분	         첫 번째 결과 (2 * std)	                            두 번째 결과 (3 * std)
코드 설정	    thr = 2 * resid.std()	                            thr = 3 * resid.std()
임계값 높이	    낮음 (더 좁은 오차 범위 허용)	                    높음 (더 넓은 오차 범위 허용)
감지 민감도	    높음 (작은 변동도 모두 감지)	                    낮음 (확실하게 큰 변동만 감지)
후반부(160~250)	자잘한 노이즈까지 수많은 점선이 빽빽하게 포착됨	    잔노이즈가 필터링되고, 시퀀스 200의 대형 급락 구간만 포착됨
초반부(30 근처)	    미세 변동 감지됨	                        여전히 감지됨 (평탄 구간의 잔차 std가 워낙 작기 때문)

## 실습 7. window·기준값 바꿔 결과 비교
### 목표
창 크기와 임계 배수를 바꿔 변화점 개수가 어떻게 달라지는지 비교
### 단계
· 잔차 기반 변화점 개수를 세는 함수를 만들기
· 창 크기를 바꿔 가며 변화점 개수 비교
· 임계 배수를 바꿔 가며 변화점 개수 비교
### 예상 결과
창이 커지거나 배수가 작아질수록 변화점 개수가 달라짐

```python
# 실습 7 코드
series = unit2["회전A전류"]

# 1. window(창 크기) 변화에 따른 이상치 개수 확인 (k=2 고정)
k = 2
for window in [5, 10, 20]:
    r = series - series.rolling(window).mean()
    print("window :", window, int((r.abs() > k * r.std()).sum()))

print("-" * 30)

# 2. k(임계값 배율) 변화에 따른 이상치 개수 확인 (window=10 고정)
window = 10
for k in [1.5, 2.0, 3.0]:
    r = series - series.rolling(window).mean()
    print("k :", k, int((r.abs() > k * r.std()).sum()))
```

## 실습 8. 단일 설비 상태 추적 미니 통합
### 목표
이동통계·변화점·변동성을 한데 모아 한 엔진의 상태를 요약
### 단계
- 이동평균·이동표준편차·변화점을 한 번에 계산
- 초반·후반 변동성과 변화점 개수를 요약
- 후반 변화점 비율로 상태를 한 문장으로 진단
### 예상 결과
후반 변동성이 커지고 변화점 13개, 후반 비율이 높아 열화 진행

```python
# 실습 8 코드

print("\n ============ 회전A전류, window = 10 ================= ")
series = unit2["회전A전류"]
window = 10

unit2["ma"] = series.rolling(window).mean() # 이동평균
unit2["std"] = series.rolling(window).std() # 이동표준편차
# unit2.head()

resid = series - unit2["ma"]
trr = 2 * resid.std()                       # 잔차 표준편차의 2배를 임계값으로 정함
unit2["chg"] = resid.abs() > thr            # 변화점 찾기
half = unit2["시퀀스"].median()             # 중간시퀀스

print("\n 전반부 변동성")
print("이동표준편차 : ",round(unit2["std"].iloc[10:50].mean(), 4))

print("\n 후반부 변동성")
print("이동표준편차 : ",round(unit2["std"].iloc[150:].mean(), 4))


print("\n --------")
print("변화점 : ", int(unit2["chg"].sum()))
print("후반 비율 : ", round(unit2.loc[unit2["시퀀스"] > half, "chg"].mean(), 2))
# 0.09로 후반부에 몰려있다 생각하면 됨


print("\n ============ 회전A전류,window = 50 ================= ")
series = unit2["회전A전류"]
window = 50

unit2["ma"] = series.rolling(window).mean() 
unit2["std"] = series.rolling(window).std() 
# unit2.head()

resid = series - unit2["ma"]
trr = 2 * resid.std()                       
unit2["chg"] = resid.abs() > thr            
half = unit2["시퀀스"].median()             

print("\n 전반부 변동성")
print("이동표준편차 : ",round(unit2["std"].iloc[10:50].mean(), 4))

print("\n 후반부 변동성")
print("이동표준편차 : ",round(unit2["std"].iloc[150:].mean(), 4))


print("\n --------")
print("변화점 : ", int(unit2["chg"].sum()))
print("후반 비율 : ", round(unit2.loc[unit2["시퀀스"] > half, "chg"].mean(), 2))

```
