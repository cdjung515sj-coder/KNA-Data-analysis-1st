# 01. Pandas 데이터프레임 기초

## 핵심 개념
- `Series` : 1차원 데이터
- `DataFrame` : 행과 열로 이루어진 2차원 표
- Pandas는 CSV 같은 표 데이터를 읽고, 확인하고, 가공하는 데 사용


## 자주 쓰는 함수·속성·메서드

| 코드 | 종류 | 의미 |
|---|---|---|
| `pd.read_csv()` | 함수 | CSV 불러오기 |
| `df.head()` | 메서드 | 앞부분 확인 |
| `df.info()` | 메서드 | 자료형·결측치 확인 |
| `df.shape` | 속성 | `(행 수, 열 수)` |
| `df.columns` | 속성 | 컬럼명 확인 |
| `df.dtypes` | 속성 | 각 컬럼 자료형 |
| `len(df)` | 함수 | 행 개수 |


````python
import pandas as pd

df = pd.read_csv("data.csv")

print(df.shape)
print(df.columns)

df.head()
````

## Series vs DataFrame

```python
df["온도"]              # Series
df[["온도"]]            # DataFrame
df[["온도","진동"]]     # DataFrame
```

### 기억
- 대괄호 1번 + 컬럼명 1개 → Series
- 대괄호 2번 → DataFrame
