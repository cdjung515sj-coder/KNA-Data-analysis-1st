# 01. Matplotlib 기초와 선 그래프

## 기본 구조

```python
import matplotlib.pyplot as plt
```

`plt`는 `matplotlib.pyplot`의 일반적인 별명입니다.


## 핵심 함수

| 코드 | 역할 |
|---|---|
| `plt.figure()` | 도화지 크기 |
| `plt.plot()` | 선 그래프 |
| `plt.title()` | 제목 |
| `plt.xlabel()` | x축 이름 |
| `plt.ylabel()` | y축 이름 |
| `plt.xticks()` | x축 눈금 |
| `plt.legend()` | 범례 |
| `plt.grid()` | 격자 |
| `plt.savefig()` | 이미지 저장 |
| `plt.show()` | 그래프 표시 |


## 가장 중요한 코드 순서

**figure → plot → 꾸미기 → savefig → show**


````python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [62,63,65,68,72]

plt.figure(figsize=(8,4))
plt.plot(x, y, marker="o", label="A호기")

plt.title("A호기 온도 변화")
plt.xlabel("일자")
plt.ylabel("온도(℃)")
plt.grid(True, alpha=0.3)
plt.legend()

plt.show()
````

## 여러 선

`plot()`을 여러 번 호출하면 같은 그래프에 겹쳐 그려집니다.

`label=`로 이름을 적고 `plt.legend()`로 범례를 표시합니다.


````python
x = [1,2,3]
a = [65,68,70]
b = [66,71,74]

plt.plot(x, a, label="A호기")
plt.plot(x, b, label="B호기")
plt.legend()
plt.show()
````
