# 02. Pandas 선택과 필터링

## 핵심
데이터 분석에서 가장 자주 하는 일은
**필요한 행과 열만 골라내는 것**입니다.


## 핵심 도구

| 코드 | 의미 |
|---|---|
| `df["열"]` | 열 선택 |
| `df[["열1","열2"]]` | 여러 열 선택 |
| `df.loc[행조건, 열]` | 이름/조건 기준 선택 |
| `df.iloc[행번호, 열번호]` | 위치 기준 선택 |
| `df[조건]` | 불리언 필터링 |
| `.isin([...])` | 여러 값 중 포함 여부 |
| `.between(a,b)` | a~b 범위 |
| `~조건` | 조건 반대 |


````python
# 예시 데이터
import pandas as pd

df = pd.DataFrame({
    "설비": ["A","B","C","D"],
    "온도": [70, 82, 76, 91],
    "진동": [2.1, 3.7, 2.8, 5.1],
    "상태": ["정상","점검","정상","점검"]
})

df
````

## 조건식

```python
df["온도"] > 80
```

결과는 `True / False`로 이루어진 불리언 Series입니다.

이 조건을 DataFrame에 넣으면 True인 행만 남습니다.


````python
df[df["온도"] > 80]
````

## 여러 조건

```python
(df["온도"] > 80) & (df["진동"] > 3)
(df["온도"] > 80) | (df["진동"] > 3)
```

- `&` : AND
- `|` : OR
- 각 조건은 반드시 괄호로 묶기


````python
df[(df["온도"] > 80) & (df["진동"] > 3)]
````

## isin / between / 부정

```python
df[df["상태"].isin(["점검", "고장"])]
df[df["온도"].between(70, 80)]
df[~df["상태"].isin(["정상"])]
```
