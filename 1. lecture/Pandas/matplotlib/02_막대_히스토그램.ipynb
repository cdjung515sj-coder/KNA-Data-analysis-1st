## 막대그래프 `plt.bar()`

범주 비교에 사용합니다.

```python
plt.bar(항목, 값)
```


````python
import matplotlib.pyplot as plt

names = ["A호기","B호기","C호기"]
temps = [64,70,77]

plt.bar(names, temps, width=0.6)
plt.ylabel("평균 온도(℃)")
plt.title("설비별 평균 온도")
plt.show()
````

### 주요 옵션
- `color=` : 색
- `width=` : 막대 굵기
- `plt.barh()` : 가로 막대
- `plt.xticks(rotation=45)` : 긴 글자 기울이기

### 주의
막대그래프는 크기 비교가 목적이므로 보통 y축을 0부터 시작합니다.


## 히스토그램 `plt.hist()`

숫자 하나의 **분포**를 봅니다.

```python
plt.hist(데이터, bins=10)
```


````python
temps = [62,63,65,65,66,68,69,70,70,71,72,75,78]

plt.hist(temps, bins=5)
plt.xlabel("온도(℃)")
plt.ylabel("개수")
plt.title("온도 분포")
plt.show()
````

## bins

- 작음 → 큰 모양만 보임
- 큼 → 더 세밀함
- 너무 크면 빈 구간이 많아질 수 있음
- 정답은 없고 여러 값을 비교해서 보는 것이 좋음

### 분포 해석 순서
1. 중심 : 어디에 모였나?
2. 퍼짐 : 얼마나 흩어졌나?
3. 동떨어진 값 : 이상치 후보가 있나?

연결:
- 중심 ≈ 평균
- 퍼짐 ≈ 표준편차
- 동떨어진 값 ≈ 이상치 후보
