# 03. 산점도 · 기준선 · subplot

## 산점도 `plt.scatter()`

두 숫자형 변수의 관계를 확인합니다.

```python
plt.scatter(x, y)
```


````python
import matplotlib.pyplot as plt

temp = [65,68,70,72,75,78]
vib = [2.0,2.1,2.4,2.6,3.0,3.5]

plt.scatter(temp, vib, s=40, alpha=0.6)
plt.xlabel("온도(℃)")
plt.ylabel("진동(mm/s)")
plt.title("온도와 진동의 관계")
plt.show()
````

## 관계 읽기

- 우상향 ↗ : 양의 관계
- 우하향 ↘ : 음의 관계
- 방향 없이 흩어짐 : 뚜렷한 관계 없음

### 주의
**상관관계 ≠ 인과관계**

같이 움직인다고 해서 한쪽이 다른 쪽의 원인이라는 뜻은 아닙니다.


## 산점도 옵션

| 옵션 | 의미 |
|---|---|
| `color=` | 점 색 |
| `s=` | 점 크기 |
| `alpha=` | 투명도 |


## 기준선

```python
plt.axhline(y=4.0)   # 가로 기준선
plt.axvline(x=80)    # 세로 기준선
```


````python
plt.scatter(temp, vib, alpha=0.6)
plt.axhline(y=3.0, linestyle="--")
plt.xlabel("온도(℃)")
plt.ylabel("진동(mm/s)")
plt.show()
````

## subplot

한 화면에 여러 그래프를 배치합니다.

```python
plt.subplot(행, 열, 칸번호)
```

칸 번호는 1부터 시작하며 왼쪽→오른쪽, 위→아래 순서입니다.


````python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.hist(temp, bins=4)
plt.title("온도 분포")

plt.subplot(1,2,2)
plt.scatter(temp, vib)
plt.title("온도-진동 관계")

plt.suptitle("설비 센서 분석")
plt.tight_layout()
plt.show()
````
