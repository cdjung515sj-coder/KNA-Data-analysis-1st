# 03. 정렬 · groupby · 집계

## 분석 흐름
**불러오기 → 확인 → 필터링 → 정렬 → 집계**


## 핵심 메서드

| 코드 | 의미 |
|---|---|
| `sort_values()` | 값 기준 정렬 |
| `value_counts()` | 값별 개수 |
| `groupby()` | 그룹 나누기 |
| `mean()` | 평균 |
| `sum()` | 합계 |
| `min()` / `max()` | 최소/최대 |
| `count()` | 결측치 제외 개수 |
| `size()` | 행 개수 |
| `agg()` | 여러 집계를 한 번에 |
| `round()` | 소수점 정리 |
| `var()` | 분산 |
| `std()` | 표준편차 |


````python
import pandas as pd

df = pd.DataFrame({
    "설비라인": ["A","A","B","B","C","C"],
    "온도": [70,72,75,77,80,82],
    "진동": [2.0,2.2,2.5,2.8,3.1,3.4],
    "판정": ["정상","정상","정상","점검","점검","점검"]
})
````

## 정렬

```python
df.sort_values("온도")
df.sort_values("온도", ascending=False)
```


````python
df.sort_values("온도", ascending=False)
````

## value_counts

```python
df["판정"].value_counts()
```

범주별 개수를 빠르게 확인할 때 사용합니다.


````python
df["판정"].value_counts()
````

## groupby

```python
df.groupby("설비라인")["온도"].mean()
```

뜻:
1. `설비라인`으로 그룹을 나누고
2. 각 그룹의 `온도`를 선택하고
3. 평균을 구한다


````python
df.groupby("설비라인")["온도"].mean()
````

## agg

여러 통계량을 한 번에 구할 때 사용합니다.


````python
df.groupby("설비라인").agg(
    온도평균=("온도", "mean"),
    진동평균=("진동", "mean"),
    측정수=("온도", "count")
).round(2)
````

## size vs count

- `size()` : 행 자체의 개수
- `count()` : 결측치가 아닌 값의 개수

결측치가 있으면 결과가 달라질 수 있습니다.
