## 핵심
전처리는 값을 무조건 지우는 작업이 아니라
**왜 이상한지 확인하고, 처리 후 결과를 다시 검증하는 과정**입니다.


## 결측치 기본

| 코드 | 의미 |
|---|---|
| `df.isna()` | 결측 여부 |
| `df.isna().sum()` | 컬럼별 결측 개수 |
| `df.dropna()` | 결측 행 제거 |
| `df.fillna()` | 결측값 채우기 |


````python
# 예시
df["온도"].isna().sum()
````

## 이상치 1: IQR

사분위수를 이용하는 방법.

```text
IQR = Q3 - Q1

하한 = Q1 - 1.5 × IQR
상한 = Q3 + 1.5 × IQR
```

하한보다 작거나 상한보다 큰 값을 이상치 후보로 봅니다.


````python
q1 = df["온도"].quantile(0.25)
q3 = df["온도"].quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outlier = df[(df["온도"] < lower) | (df["온도"] > upper)]
outlier
````

## 이상치 2: 3시그마

평균에서 표준편차 3배 이상 떨어진 값을 이상치 후보로 보는 방식.

```text
평균 - 3×표준편차 ~ 평균 + 3×표준편차
```

정규분포에 가까운 데이터에서 많이 사용합니다.


## 꼭 기억
- 이상치 = 무조건 오류가 아님
- 실제 고장 신호일 수도 있음
- 처리 전/후의 분포를 그래프로 비교하면 좋음
